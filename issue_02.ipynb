```python
import sqlite3

def criar_tabela_medicoes():
    conexao = sqlite3.connect("tanques.db")
    cursor = conexao.cursor()
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS medicoes (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            tanque_nome TEXT NOT NULL,
            nivel REAL NOT NULL,
            data_hora TEXT DEFAULT CURRENT_TIMESTAMP
        )
    """)
    conexao.commit()
    conexao.close()

def salvar_medicao(tanque_nome, nivel):
    conexao = sqlite3.connect("tanques.db")
    cursor = conexao.cursor()
    cursor.execute(
        "INSERT INTO medicoes (tanque_nome, nivel) VALUES (?, ?)",
        (tanque_nome, nivel)
    )
    conexao.commit()
    conexao.close()
```

```python
import matplotlib.pyplot as plt
import numpy as np


class Tanque:
    def __init__(self, nome, capacidade):
        self.nome = nome
        self.capacidade = capacidade
        self.nivel_atual = 0

        # Vetor de medições simuladas
        self.conjunto = np.array([650, 720, 810, 780, 690])

    def atualizar_nivel(self, novo_nivel):
        """Atualiza o nível quando há uma nova medição"""
        self.nivel_atual = novo_nivel
         salvar_medicao(self.nome, novo_nivel)
        print(f"Nova medição registrada: {self.nivel_atual} L")

    def mostrar_nivel(self):
        print(f"Nível atual do tanque: {self.nivel_atual} L")

    def simular(self):
        for nivel in self.conjunto:
            self.atualizar_nivel(nivel)

        tempo = np.arange(1, len(self.conjunto) + 1)

        plt.plot(tempo, self.conjunto, marker="o")
        plt.title(f"Nível de água - {self.nome}")
        plt.xlabel("Medição")
        plt.ylabel("Nível (L)")
        plt.grid(True)
        plt.show()


# Teste
tanque = Tanque("Caixa d'água", 10000)

tanque.mostrar_nivel()
tanque.simular()
```
