```python
import sqlite3

conexao = sqlite3.connect("tanque.db")
cursor = conexao.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS tanque(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    nome TEXT NOT NULL,
    capacidade REAL NOT NULL
)
""")

conexao.commit()
```

```python
import matplotlib.pyplot as plt
import numpy as np

class Tanque:
    def __init__(self, nome, capacidade):
        self.nome = nome
        self.capacidade = capacidade
        self.nivel_atual = 0

        # Vetor de medições
        self.conjunto = np.array([650, 720, 810, 780, 690])

    def atualizar_nivel(self, novo_nivel):
        self.nivel_atual = novo_nivel
        print(f"Nova medição: {self.nivel_atual} L")

    def mostrar_nivel(self):
        print(f"Nível atual: {self.nivel_atual} L")

    def simular(self):
        for nivel in self.conjunto:
            self.atualizar_nivel(nivel)

        tempo = np.arange(1, len(self.conjunto) + 1)

        plt.plot(tempo, self.conjunto, marker="o")
        plt.title("Nível de Água do Tanque")
        plt.xlabel("Medição")
        plt.ylabel("Nível (L)")
        plt.grid(True)
        plt.show()


tanque = Tanque("Caixa d'água", 10000)

tanque.mostrar_nivel()
tanque.simular()
```

```python
class Tanque:
    def __init__(self, nome, capacidade):
        self.nome = nome
        self.capacidade = capacidade
        self.nivel_atual = 0

    def atualizar_nivel(self, novo_nivel):
        self.nivel_atual = novo_nivel
        print(f"Nova medição registrada: {self.nivel_atual} litros")

    def mostrar_nivel(self):
        print(f"Nível atual do tanque: {self.nivel_atual} litros")


# Teste
tanque = Tanque("Caixa d'água", 10000)

tanque.mostrar_nivel()

tanque.atualizar_nivel(650)
tanque.mostrar_nivel()

tanque.atualizar_nivel(720)
tanque.mostrar_nivel()

tanque.atualizar_nivel(810)
tanque.mostrar_nivel()
```
